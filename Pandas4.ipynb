```python
import pandas as pd
import numpy as np
movies = pd.read_csv('movies.csv', index_col=0)
directors = pd.read_csv('directors.csv',index_col=0)
data = movies.merge(directors, how='left', left_on='director_id',right_on='id')
data.drop(['director_id','id_y'],axis=1,inplace=True)
```

```python
data["revenue"]=(data["revenue"]/1000000).round(2)
data["budget"]=(data["budget"]/1000000).round(2)
```

```python
#custom functions

def temp_mean(x):
    x["Temp_avg"]=np.round(x["Temperature"].mean(),2)
    return x
```

```python
df_tidy = df_tidy.groupby("Drug_Name").apply(temp_mean)
df_tidy
```

```python
df_tidy
```

```python
#ignore warnings code
import warnings
warnings.filterwarnings("ignore")

df_tidy=df_tidy.groupby("Drug_Name").apply(temp_mean)
df_tidy
```

```python
df_tidy["Temperature"].fillna(df_tidy["Temp_avg"],inplace=True)
```

```python
df_tidy
```

```python
def pres_mean(x):
    x["Pres_avg"]=np.round(x["Pressure"].mean(),2)
    return x
```

```python
df_tidy=df_tidy.groupby("Drug_Name").apply(pres_mean)

df_tidy["Pressure"].fillna(df_tidy["Pres_avg"],inplace=True)
df_tidy
```
